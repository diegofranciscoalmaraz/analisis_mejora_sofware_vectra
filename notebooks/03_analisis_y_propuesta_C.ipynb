### Propósito del documento

Analizar la organización conceptual del software de VECTRA e identificar oportunidades de mejora mediante una propuesta de arquitectura basada en etapas independientes para el procesamiento de datos de telemetría.

Esta propuesta busca establecer responsabilidades claras para la recepción, validación, normalización, almacenamiento y análisis de las mediciones obtenidas del vehículo, favoreciendo un diseño comprensible, mantenible y susceptible de ampliaciones futuras.

**Alcance:** El análisis se basa en la arquitectura conceptual del proyecto. Los componentes y clases descritos en la propuesta representan alternativas de diseño y no deben interpretarse como funcionalidades ya implementadas.

## 1. Introducción

VECTRA es un proyecto conceptual de monitoreo vehicular inteligente cuyo propósito es recopilar y analizar información relacionada con las condiciones de funcionamiento y el desplazamiento de un vehículo. Para ello, se contempla el uso de sensores, un microcontrolador ESP32, comunicación inalámbrica y una plataforma para consultar los datos registrados.

La información de telemetría puede incluir variables como aceleración, vibración, inclinación, temperatura y ubicación geográfica. Su procesamiento permitiría identificar comportamientos relevantes, conservar registros históricos y generar alertas ante posibles situaciones anómalas, de acuerdo con las reglas que se definan para el sistema.

Desde la perspectiva de la ingeniería de software, la organización de estos componentes es importante para establecer responsabilidades claras y facilitar el mantenimiento y la evolución del sistema. Una arquitectura poco definida podría dificultar la modificación de una función sin afectar otras partes del proceso.

En este contexto, el presente documento analiza la arquitectura conceptual de VECTRA y desarrolla una propuesta de mejora centrada en organizar el procesamiento de la telemetría mediante etapas con responsabilidades específicas.

### 1.1 Objetivo general

Proponer una organización del procesamiento de telemetría que permita separar las responsabilidades de cada etapa y favorezca la mantenibilidad, la capacidad de prueba y la incorporación de nuevas funcionalidades.

### 1.2 Alcance del análisis

El documento presenta una evaluación conceptual del diseño, identifica posibles riesgos de organización y describe una alternativa arquitectónica. Las dificultades señaladas se consideran riesgos potenciales que deberán verificarse frente a una implementación real; asimismo, las clases propuestas representan elementos de diseño y no necesariamente componentes existentes en el proyecto.

## 2. Análisis de la arquitectura conceptual

La arquitectura conceptual de VECTRA puede organizarse en componentes que participan en la captura, transmisión, procesamiento y presentación de la información vehicular. Cada componente cumple una función específica dentro del flujo de datos.

La siguiente tabla resume las responsabilidades consideradas en el diseño conceptual del proyecto. Esta descripción no confirma la existencia de módulos o clases implementados; su finalidad es establecer una base para evaluar la organización del software.

### 2.1 Componentes y responsabilidades

### 2.2 Interpretación del flujo de información

Desde una perspectiva general, el flujo comienza con la captura de mediciones mediante los sensores. El ESP32 participa en la recopilación y transmisión de estos datos; posteriormente, los servicios de recepción, procesamiento y almacenamiento permiten preparar y conservar la información.

A partir de los datos disponibles, el sistema podría realizar análisis para detectar comportamientos inusuales y generar alertas cuando se cumplan las condiciones establecidas. Finalmente, el dashboard permitiría consultar los resultados y revisar los registros históricos.

Esta secuencia representa una organización conceptual del sistema. El orden exacto de algunas operaciones, así como la relación entre el almacenamiento y el análisis, dependerá de las decisiones de implementación.

### 2.3 Aspectos relevantes para el diseño

La separación de responsabilidades es un criterio importante para evaluar la calidad del diseño. Si la recepción, la validación, el almacenamiento y el análisis se concentran en un mismo componente, las modificaciones podrían resultar más difíciles de probar y mantener.

Por ello, conviene analizar si cada actividad puede definirse mediante una responsabilidad concreta y si sus relaciones con los demás componentes están suficientemente delimitadas. Este análisis permitirá fundamentar la propuesta de mejora que se presenta en las siguientes secciones.


## 3. Identificación de riesgos potenciales del diseño

A partir de la arquitectura conceptual descrita, es posible identificar algunos riesgos que conviene evaluar antes de implementar o ampliar el sistema. Estos riesgos no constituyen defectos comprobados del software, ya que su confirmación requiere revisar el código y el comportamiento de una implementación real.

### 3.1 Delimitación insuficiente de responsabilidades

Si un componente se encarga simultáneamente de recibir, validar, transformar y analizar la telemetría, sus responsabilidades pueden resultar difíciles de distinguir. Esto aumentaría la complejidad de comprender y modificar el sistema.

**Posible consecuencia:** un cambio en una operación podría requerir modificar otras funciones relacionadas.

### 3.2 Dependencia entre etapas del procesamiento

Cuando las actividades del sistema están estrechamente relacionadas, una modificación en el formato de los datos o en una regla de análisis podría afectar a otros componentes.

**Posible consecuencia:** el mantenimiento y la incorporación de nuevas funciones podrían requerir más ajustes de los necesarios.

### 3.3 Dificultad para realizar pruebas independientes

Si la validación, la normalización y el análisis dependen directamente unos de otros, probar cada operación de manera aislada puede resultar más complicado.

**Posible consecuencia:** sería más difícil identificar en qué etapa se origina un error y comprobar cada comportamiento por separado.

### 3.4 Tratamiento inconsistente de los datos

Si no se establecen criterios claros para validar y normalizar las mediciones, distintos componentes podrían interpretar de manera diferente datos incompletos, inválidos o expresados en formatos distintos.

**Posible consecuencia:** los resultados del análisis podrían perder consistencia o requerir correcciones posteriores.

### 3.5 Dificultad para ampliar el procesamiento

La incorporación de nuevas operaciones, como filtrar lecturas duplicadas o enriquecer los datos con información del recorrido, podría exigir cambios en componentes existentes si no hay una separación adecuada de responsabilidades.

**Posible consecuencia:** el crecimiento del sistema podría incrementar su complejidad y el esfuerzo de mantenimiento.

### 3.6 Síntesis del análisis

Los riesgos anteriores señalan una oportunidad de mejora: organizar el procesamiento de telemetría en etapas con funciones definidas y relaciones explícitas. Esta alternativa permitiría evaluar cada operación de forma independiente y reducir las dependencias innecesarias.

No obstante, la conveniencia de aplicar esta estructura deberá valorarse según la complejidad real del sistema. Una arquitectura más modular puede aportar beneficios, pero también introduce componentes y tareas de coordinación adicionales.

## 5. Propuesta de mejora: procesamiento de telemetría por etapas

### 5.1 Descripción de la propuesta

Se propone organizar el procesamiento de la telemetría de VECTRA mediante una secuencia de etapas independientes, cada una con una responsabilidad claramente definida. El propósito es reducir la dependencia entre operaciones y facilitar la comprensión, las pruebas y el mantenimiento del sistema.

En lugar de tratar el procesamiento como una actividad general, la propuesta distingue las operaciones necesarias para preparar los datos y utilizarlos posteriormente en el análisis.

### 5.2 Etapas propuestas

**1. Recepción de telemetría**

Recibir las mediciones enviadas por el dispositivo y entregarlas al proceso de tratamiento de datos. Esta etapa no debería asumir responsabilidades de análisis de anomalías.

**2. Validación de datos**

Comprobar que los datos recibidos contengan los campos requeridos y cumplan las reglas básicas de validez. Las mediciones incompletas o incorrectas deberán tratarse de acuerdo con criterios definidos por el sistema.

**3. Normalización**

Convertir los datos a una representación uniforme. Por ejemplo, se podrían estandarizar nombres de campos, unidades de medida y formatos de fecha para facilitar su utilización posterior.

**4. Enriquecimiento del contexto**

Incorporar información complementaria cuando esté disponible, como el identificador del vehículo, la fecha de la medición o el recorrido asociado. Esta etapa dependerá de los datos y servicios que se definan en la implementación.

**5. Almacenamiento**

Conservar las mediciones procesadas para permitir su consulta y utilización posterior. La estrategia de persistencia deberá establecerse de acuerdo con los requisitos del proyecto.

**6. Análisis de anomalías**

Evaluar las mediciones mediante reglas o criterios previamente definidos para identificar comportamientos que requieran atención. La detección efectiva dependerá de los algoritmos y umbrales que se implementen.

**7. Gestión de alertas**

Recibir los resultados del análisis y determinar si corresponde generar una notificación. La gestión de alertas deberá distinguir entre la detección de un evento y su comunicación al usuario.

### 5.3 Criterio de diseño

La propuesta se fundamenta en la separación de responsabilidades: cada etapa debe realizar una tarea específica y comunicar su resultado a la siguiente mediante un formato de datos acordado.

Esta organización puede facilitar las pruebas unitarias, porque cada etapa podría evaluarse con entradas y resultados controlados. También permitiría incorporar operaciones adicionales, como la eliminación de lecturas duplicadas, sin tener que rediseñar necesariamente todo el proceso.

La secuencia descrita es una propuesta arquitectónica. Su implementación requerirá definir los contratos entre etapas, el manejo de errores, el tratamiento de datos inválidos y las condiciones bajo las cuales se generan las alertas.

### 6.2 Interpretación del diseño

La arquitectura propuesta organiza el procesamiento como una secuencia de operaciones. Primero se reciben las mediciones; después se verifica su validez, se uniforman sus formatos y se incorpora contexto adicional cuando esté disponible.

Una vez preparadas, las mediciones se almacenan y quedan disponibles para el análisis de anomalías. Si se identifica una condición que cumple los criterios establecidos, la etapa de gestión de alertas determina la acción correspondiente.

El diagrama también contempla el tratamiento de datos inválidos. La estrategia concreta —por ejemplo, registrar el error, rechazar la medición o enviarla a revisión— deberá definirse durante la implementación.

Asimismo, será necesario definir qué ocurre cuando una etapa falla, cómo se registran los errores y si determinadas operaciones deben ejecutarse de forma síncrona o asíncrona. Estas decisiones forman parte del diseño detallado y no quedan resueltas únicamente mediante el diagrama.

**Nota:** los componentes representados son elementos propuestos para organizar el software; su presencia en el sistema implementado deberá verificarse.

## 7. Componentes y clases propuestos

Para trasladar la arquitectura conceptual a un diseño orientado a objetos, se propone definir clases con responsabilidades específicas. Esta separación busca evitar que un único componente concentre todas las operaciones del procesamiento de telemetría.

Las siguientes clases son una propuesta de diseño; no se afirma que existan actualmente en el código de VECTRA.

### 7.1 Responsabilidades de las clases

| Clase propuesta | Responsabilidad | Entrada principal | Resultado esperado |
|---|---|---|---|
| `TelemetryReceiver` | Recibir las mediciones procedentes del dispositivo o del servicio de comunicación. | Mensajes de telemetría. | Datos recibidos. |
| `TelemetryPipeline` | Coordinar la ejecución de las etapas de procesamiento. | Datos recibidos. | Flujo de procesamiento coordinado. |
| `TelemetryValidator` | Verificar la estructura y las reglas básicas de validez de los datos. | Datos recibidos. | Datos válidos o resultado de validación. |
| `TelemetryNormalizer` | Uniformar formatos, nombres de campos y unidades cuando corresponda. | Datos validados. | Datos normalizados. |
| `TelemetryEnricher` | Incorporar contexto adicional disponible, como identificadores o información del recorrido. | Datos normalizados y contexto disponible. | Datos enriquecidos. |
| `TelemetryStore` | Persistir y recuperar las mediciones procesadas. | Datos preparados para almacenamiento. | Registros almacenados o recuperados. |
| `AnomalyAnalyzer` | Evaluar las mediciones según reglas o criterios definidos. | Datos disponibles para análisis. | Resultado del análisis. |
| `AlertDispatcher` | Gestionar la preparación y distribución de las alertas que correspondan. | Resultado del análisis. | Notificación preparada o enviada. |

### 7.2 Criterios para definir las clases

La propuesta se apoya en tres criterios de diseño:

- **Responsabilidad única:** cada clase debe concentrarse en una función principal claramente delimitada.
- **Bajo acoplamiento:** las clases deberían depender lo menos posible de los detalles internos de otras clases.
- **Facilidad de prueba:** cada operación debería poder evaluarse con entradas controladas y resultados esperados.

Por ejemplo, las reglas de validación podrían probarse sin necesidad de conectar físicamente los sensores del vehículo. De forma similar, el análisis de anomalías podría evaluarse con conjuntos de datos de prueba, siempre que su lógica se defina de manera independiente.

### 7.3 Consideraciones de implementación

Para mantener una separación efectiva, las clases deberían intercambiar datos mediante estructuras definidas y evitar compartir responsabilidades innecesarias. También convendría establecer mecanismos explícitos para comunicar errores, registrar fallos y gestionar las operaciones de almacenamiento.

La clase `TelemetryPipeline` actuaría como coordinadora del flujo, pero no debería absorber las tareas específicas de validación, normalización o análisis. De esta manera, las etapas podrían modificarse o probarse individualmente sin concentrar toda la lógica en un único componente.

## 9. Justificación de la propuesta

La propuesta C se justifica por la necesidad de definir con mayor precisión las responsabilidades que intervienen en el procesamiento de la telemetría. La separación en etapas permite distinguir la recepción de los datos, su validación, la normalización, el almacenamiento y el análisis, evitando que todas estas operaciones dependan de un único componente con múltiples funciones.

Además, la coordinación del flujo mediante un componente específico permite representar el orden de las operaciones de forma explícita. Esta organización facilita comprender el proceso completo y establecer criterios de entrada, salida y manejo de errores para cada etapa.

La propuesta no garantiza por sí misma una mejora del rendimiento ni elimina todos los defectos posibles. Sus beneficios dependerán de una implementación consistente, de la definición de contratos entre componentes y de las pruebas que permitan verificar su comportamiento.

## 10. Beneficios esperados y limitaciones

### 10.1 Beneficios esperados

- **Mayor claridad del diseño:** cada etapa tiene una responsabilidad identificable, lo que facilita comprender el recorrido de la telemetría.
- **Pruebas más específicas:** las operaciones podrían verificarse individualmente mediante datos de prueba controlados.
- **Mantenimiento más organizado:** los cambios en una etapa podrían limitarse a su responsabilidad si las dependencias están bien definidas.
- **Ampliación del procesamiento:** podrían incorporarse operaciones adicionales, como el filtrado de lecturas duplicadas, cuando exista una necesidad concreta.
- **Mejor gestión de errores:** la separación permite definir cómo se detectan, registran y tratan los problemas en cada etapa.

### 10.2 Limitaciones y consideraciones

La arquitectura por etapas también presenta desafíos. Cada componente adicional requiere definiciones, pruebas y mantenimiento, y la coordinación del flujo introduce cierta complejidad.

Si el procesamiento real fuera pequeño y estable, una estructura demasiado fragmentada podría resultar innecesaria. También sería importante evitar que los datos se transformen repetidamente o que los errores se pierdan al pasar de una etapa a otra.

Por ello, la propuesta debe aplicarse de manera proporcional a las necesidades de VECTRA. La modularidad es útil cuando mejora la comprensión, la prueba o la evolución del sistema; no constituye un objetivo independiente.
