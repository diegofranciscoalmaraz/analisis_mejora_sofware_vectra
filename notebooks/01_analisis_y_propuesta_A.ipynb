## Objetivo

Analizar el diseño actual de VECTRA para identificar posibles problemas en la organización de sus componentes y responsabilidades, y elaborar una propuesta de mejora que favorezca la modularidad, la mantenibilidad y la facilidad de ampliación del sistema.

## Alcance

Este análisis se centra en la arquitectura y las responsabilidades de los componentes del sistema. Las observaciones se distinguirán entre características documentadas, problemas comprobables y posibles riesgos de diseño.

La propuesta se presentará como una alternativa de diseño, sin asumir que las clases propuestas ya existen en la implementación actual.

## 1. Descripción general del sistema

VECTRA es un sistema IoT orientado al monitoreo de vehículos mediante la recopilación y el análisis de telemetría. El sistema contempla variables como aceleración, vibración, inclinación, temperatura, velocidad y ubicación GPS.

La información recopilada permite analizar el comportamiento del vehículo y detectar posibles anomalías al comparar las mediciones con los patrones de funcionamiento habituales.

### 1.1 Arquitectura conceptual

La arquitectura descrita para VECTRA se organiza en cuatro capas:

- **Sensado:** obtiene las mediciones del vehículo mediante sensores.
- **Comunicación:** permite transmitir la información recopilada.
- **Datos y análisis:** procesa la telemetría y analiza el comportamiento del vehículo.
- **Aplicación:** presenta información relevante para el usuario.

### 1.2 Propósito del análisis

A partir de esta arquitectura, se estudiará cómo se distribuyen las responsabilidades entre los componentes y qué alternativas permitirían reducir el acoplamiento, facilitar las pruebas y simplificar futuras ampliaciones.

**Nota sobre el alcance de la información:** esta descripción representa la arquitectura conceptual disponible. Para confirmar las clases implementadas, sus métodos y sus dependencias, será necesario revisar el código fuente o los diagramas detallados del proyecto.

## 2. Análisis del diseño actual

### 2.1 Estructura arquitectónica identificada

De acuerdo a lo que propusimos en *VECTRA — IoT, Datos e IA*, la arquitectura conceptual se organiza en cuatro capas:

| Capa | Responsabilidad descrita |
|---|---|
| Capa 1 — Sensado | Obtener mediciones de aceleración, vibración, inclinación, temperatura y GPS. |
| Capa 2 — Comunicación | Transmitir los datos desde el ESP32 hacia la plataforma mediante Wi-Fi. |
| Capa 3 — Datos y análisis | Almacenar y procesar los datos, además de incorporar análisis mediante inteligencia artificial. |
| Capa 4 — Aplicación | Presentar información mediante un dashboard, gráficas, alertas e históricos. |

Esta separación permite distinguir la adquisición física de datos, su transmisión, su procesamiento y la presentación de resultados.

### 2.2 Responsabilidades funcionales identificadas

Además de las cuatro capas, se contempla las siguientes funcionalidades principales:

- Registrar los recorridos del vehículo.
- Enviar o almacenar temporalmente las mediciones.
- Visualizar gráficas de las variables registradas.
- Consultar y comparar recorridos históricos.
- Detectar comportamientos que se desvíen de los patrones habituales.
- Generar alertas ante anomalías relevantes.
- Mostrar el trayecto cuando se dispone de información GPS.
- Generar reportes de los recorridos.

Estas funcionalidades permiten identificar posibles responsabilidades de software que deberán distribuirse entre componentes bien definidos en una futura implementación.

### 2.3 Evaluación del diseño documentado

La arquitectura propuesta tiene como fortaleza inicial la separación de responsabilidades por capas. Esta organización puede facilitar la evolución del sistema porque permite estudiar de manera diferenciada la adquisición, la comunicación, el análisis y la interfaz de usuario.

Sin embargo, el documento describe el diseño a nivel conceptual. No especifica todavía las clases de software, sus métodos, sus interfaces ni las dependencias concretas entre ellas. Por tanto, no es posible concluir únicamente a partir de este documento que una clase específica tenga demasiadas responsabilidades o que exista acoplamiento excesivo en el código.

La principal oportunidad de mejora que se observa en esta etapa es **definir con mayor precisión cómo se distribuyen las responsabilidades dentro de cada capa y cómo se comunican sus componentes**.

### 2.4 Alcance de las observaciones

Para mantener un análisis técnicamente correcto, se distinguen dos tipos de observaciones:

- **Hechos documentados:** la arquitectura contempla cuatro capas y funcionalidades relacionadas con recorridos, telemetría, almacenamiento, análisis, visualización y alertas.
- **Aspectos pendientes de definición:** la distribución interna de responsabilidades, las interfaces entre componentes, el manejo de errores de comunicación y la organización de los módulos de análisis.

Los aspectos pendientes representan oportunidades de diseño que deberán resolverse en la propuesta de mejora; no constituyen, por sí mismos, defectos comprobados de una implementación existente.

## 3. Identificación de problemas y oportunidades de mejora

A partir de la arquitectura y las funcionalidades descritas en el documento de VECTRA, se identifican varios aspectos que podrían definirse con mayor precisión para facilitar el desarrollo y mantenimiento del software.

### 3.1 Responsabilidades agrupadas en la capa de datos y análisis

La tercera capa contempla almacenamiento, procesamiento e inteligencia artificial. Aunque estas actividades están relacionadas, tienen objetivos diferentes y podrían requerir componentes independientes.

**Riesgo de diseño:** si estas responsabilidades se concentran en un único componente durante la implementación, podría resultar más difícil modificar el almacenamiento sin afectar el procesamiento o cambiar el método de detección de anomalías.

**Oportunidad de mejora:** definir responsabilidades diferenciadas para persistencia de datos, procesamiento de telemetría y detección de anomalías.

**Evidencia documental:** la arquitectura conceptual agrupa almacenamiento, procesamiento e IA en la capa 3.

### 3.2 Responsabilidades de la aplicación poco detalladas

La cuarta capa contempla dashboard, gráficas, alertas e históricos. El documento también incluye la consulta de recorridos y la generación de reportes.

**Riesgo de diseño:** si la lógica de presentación, la generación de alertas y la consulta de históricos se mezclan en un mismo componente, podrían aumentar las dependencias entre funcionalidades.

**Oportunidad de mejora:** separar la presentación visual de las operaciones que consultan recorridos, preparan reportes o gestionan alertas.

**Evidencia documental:** la capa 4 agrupa distintas funciones de interacción y presentación, pero el documento no especifica sus clases ni sus interfaces internas.

### 3.3 Comunicación entre el dispositivo y la plataforma

El documento propone utilizar un ESP32 y Wi-Fi para transmitir mediciones cuando exista conectividad disponible. También contempla la posibilidad de almacenar temporalmente los datos para su posterior procesamiento.

**Aspecto pendiente de diseño:** no se detallan el mecanismo de reintento, la gestión de mediciones pendientes, la detección de registros duplicados ni la recuperación ante interrupciones de conexión.

**Oportunidad de mejora:** definir un componente de comunicación con responsabilidades claras para enviar datos y gestionar los resultados de la transmisión. Si se requiere funcionamiento sin conexión, también deberá especificarse cómo se almacenan y reenvían las mediciones pendientes.

Estos puntos son requisitos por definir, no fallos comprobados de una implementación.

### 3.4 Organización y validación de las mediciones

Se propone registros con campos como `timestamp`, `lat`, `lon`, `acc_x`, `acc_y`, `acc_z`, `vibration`, `tilt`, `temperature`, `speed` y `trip_id`.

**Aspecto pendiente de diseño:** no se especifican completamente las reglas de validación, las unidades de medida, la frecuencia de muestreo ni el tratamiento de datos incompletos o inválidos.

**Oportunidad de mejora:** definir una estructura de telemetría común y un proceso de validación antes de almacenar o analizar las mediciones. Esto permitiría mantener criterios consistentes entre la adquisición, el procesamiento y la visualización.

### 3.5 Dependencias entre las capas

La arquitectura separa sensado, comunicación, datos y análisis, y aplicación. Sin embargo, no define interfaces ni contratos de comunicación entre ellas.

**Riesgo de diseño:** si una capa depende directamente de los detalles internos de otra, un cambio tecnológico podría requerir modificaciones en varios componentes.

**Oportunidad de mejora:** establecer interfaces claras para intercambiar mediciones, consultar recorridos y comunicar resultados de análisis. Así, cada capa podría evolucionar con menor dependencia de la implementación interna de las demás.

### 3.6 Resumen de hallazgos

| Aspecto | Situación documentada | Mejora propuesta |
|---|---|---|
| Datos y análisis | Almacenamiento, procesamiento e IA se agrupan en una capa. | Separar sus responsabilidades internas. |
| Aplicación | Incluye dashboard, gráficas, alertas e históricos. | Distinguir presentación y operaciones de negocio. |
| Comunicación | Se propone ESP32 y Wi-Fi. | Especificar errores, reintentos y recuperación. |
| Telemetría | Se define una estructura conceptual de registros. | Establecer validaciones y reglas de datos. |
| Integración | Se identifican cuatro capas. | Definir interfaces y contratos entre ellas. |

### 3.7 Conclusión del análisis

La arquitectura conceptual de VECTRA proporciona una separación inicial adecuada por capas. La principal oportunidad de mejora consiste en detallar las responsabilidades internas de cada capa y establecer mecanismos de comunicación explícitos entre sus componentes.

Las observaciones anteriores no demuestran que el sistema tenga errores de implementación; identifican decisiones de diseño que todavía deben especificarse. Estas oportunidades servirán como fundamento para desarrollar la propuesta A en la siguiente sección.

## 4. Propuesta de mejora A

La propuesta consiste en organizar VECTRA en componentes con tareas más específicas. De esta manera, cada parte del sistema tendría una responsabilidad clara y sería más fácil hacer cambios sin afectar todo lo demás.

Por ejemplo, si en un futuro se cambia la forma de detectar anomalías, no debería ser necesario modificar también el módulo que recibe los datos de los sensores o el que muestra las gráficas.

Las siguientes clases son una propuesta de diseño; no se afirma que ya estén implementadas en el proyecto.

### 4.1 Clases propuestas y sus responsabilidades

| Clase | Responsabilidad |
|---|---|
| `TelemetryRecord` | Representar una lectura con sus datos, como fecha, ubicación, velocidad, temperatura y aceleración. |
| `SensorManager` | Obtener las mediciones de los sensores conectados al dispositivo. |
| `TelemetryValidator` | Revisar que las lecturas tengan los campos necesarios y datos válidos antes de procesarlas. |
| `CommunicationGateway` | Encargarse del envío de las lecturas desde el dispositivo hacia la plataforma. |
| `TripService` | Organizar las mediciones por viaje y gestionar su inicio y finalización. |
| `TelemetryRepository` | Guardar las lecturas y recuperar el historial de los viajes. |
| `AnomalyDetectionService` | Analizar las lecturas para identificar comportamientos fuera del patrón habitual. |
| `AlertService` | Gestionar las alertas generadas cuando se detecta una posible anomalía. |
| `DashboardService` | Preparar los datos que necesita la interfaz para mostrar historiales y resúmenes. |

### 4.2 ¿Cómo trabajarían juntas estas clases?

El proceso comenzaría cuando los sensores obtengan una medición. Esa información se organizaría en un registro y se revisaría antes de enviarla a la plataforma.

Una vez recibidos los datos, el sistema los asociaría al viaje correspondiente y los guardaría. Después, el componente de análisis podría revisar las mediciones y, si encuentra algo fuera del patrón esperado, generar un resultado para que el componente de alertas lo gestione.

Por su parte, el panel consultaría la información guardada para mostrar gráficas, recorridos e historiales. Así, la parte visual no tendría que encargarse directamente de guardar datos ni de analizar las mediciones.

### 4.3 Diagrama de clases propuesto

El siguiente diagrama representa la organización que se propone para mejorar el diseño. Es una propuesta propia basada en las funciones descritas en la documentación de VECTRA, no un diagrama extraído del código actual.

```mermaid
classDiagram
    class SensorManager {
        +readMeasurements()
    }

    class TelemetryRecord {
        +timestamp
        +location
        +acceleration
        +vibration
        +tilt
        +temperature
        +speed
        +tripId
    }

    class TelemetryValidator {
        +validate(record)
    }

    class CommunicationGateway {
        +send(record)
    }

    class TripService {
        +startTrip()
        +finishTrip()
        +registerMeasurement(record)
    }

    class TelemetryRepository {
        +save(record)
        +findByTrip(tripId)
    }

    class AnomalyDetectionService {
        +analyze(record)
    }

    class AlertService {
        +manageAlert(result)
    }

    class DashboardService {
        +getTripHistory(tripId)
        +getSummary(tripId)
    }

    SensorManager --> TelemetryRecord : obtiene datos
    SensorManager --> TelemetryValidator : revisa lecturas
    TelemetryValidator --> CommunicationGateway : entrega datos validos
    CommunicationGateway --> TripService : envia lecturas
    TripService --> TelemetryRepository : guarda datos
    TripService --> AnomalyDetectionService : solicita analisis
    AnomalyDetectionService --> AlertService : informa resultados
    DashboardService --> TelemetryRepository : consulta historial
```

### 4.4 ¿Qué mejoraría con esta propuesta?

- **Más claridad:** cada clase tendría una función principal y sería más fácil entender qué parte del sistema realiza cada tarea.
- **Cambios más sencillos:** se podría modificar el análisis de anomalías sin tener que cambiar directamente la comunicación con los sensores.
- **Mejor mantenimiento:** si aparece un problema al guardar las mediciones, sería más sencillo ubicar la parte que debe revisarse.
- **Mayor facilidad para ampliar el sistema:** se podrían añadir nuevas formas de analizar datos o consultar viajes sin tener que concentrar toda la lógica en un solo componente.

En conclusión, esta propuesta busca que VECTRA sea más fácil de mantener y ampliar. Los beneficios dependerían de que las responsabilidades se mantengan separadas y de que la comunicación entre los componentes esté bien definida.

## 5. Comparación y justificación de la mejora

En la documentación de VECTRA se plantea una arquitectura dividida en cuatro capas: sensores, comunicación, datos y análisis, y aplicación. Esta separación ayuda a entender las partes principales del sistema, pero todavía es necesario definir con más detalle las responsabilidades de los componentes que formarían cada capa.

La propuesta A conserva esa idea general y la lleva a una organización más específica mediante clases con responsabilidades definidas.

| Aspecto | Arquitectura | Propuesta A |
|---|---|---|
| Organización | Se divide en cuatro capas generales. | Se mantienen las funciones principales, pero se distribuyen en clases propuestas. |
| Manejo de lecturas | Se contempla el envío y almacenamiento de mediciones. | Se propone separar el registro de datos, su validación, el envío y el almacenamiento. |
| Gestión de viajes | Se contempla el registro de recorridos e historiales. | Se propone una clase `TripService` para organizar las operaciones relacionadas con los viajes. |
| Análisis de anomalías | Se contempla analizar los datos para detectar comportamientos fuera del patrón habitual. | Se propone `AnomalyDetectionService` para mantener el análisis separado de otras tareas. |
| Alertas e interfaz | Se consideran alertas, gráficas e historiales. | Se propone separar la gestión de alertas de las consultas para el panel. |

### 5.1 Justificación

La principal razón para proponer esta estructura es evitar que una sola clase termine realizando demasiadas tareas. Por ejemplo, si el componente encargado de mostrar las gráficas también guardara las mediciones y detectara anomalías, un cambio en cualquiera de esas funciones podría afectar a las demás.

Con la separación propuesta, cada componente tendría una función más clara. Esto también facilitaría las pruebas, porque sería posible revisar por separado la validación de datos, el almacenamiento y el análisis de anomalías.

Además, la propuesta permitiría ampliar VECTRA en el futuro. Por ejemplo, se podría ajustar el método de detección de anomalías sin cambiar directamente el módulo encargado de recuperar el historial de los viajes.

### 5.2 Limitaciones de la propuesta

Esta mejora se basa en la arquitectura descrita hasta ahora. No se puede asegurar que los problemas mencionados estén presentes en el código actual.

Antes de aplicar el diseño, sería necesario revisar el código disponible, confirmar cómo están distribuidas las responsabilidades y adaptar las clases propuestas a las necesidades reales del proyecto.

Por lo tanto, la propuesta debe entenderse como una alternativa de diseño que busca mejorar la claridad, el mantenimiento y la posibilidad de ampliar el sistema.

## 7. Conclusión Personal

Con este análisis se identificó que VECTRA tiene una arquitectura conceptual dividida en cuatro capas, lo que permite organizar las principales funciones del sistema. Sin embargo, todavía es necesario definir con mayor claridad las responsabilidades internas de cada componente y la forma en que se comunican entre sí.

Mi propuesta consiste en separar esas responsabilidades en clases especializadas, como `TelemetryValidator`, `TelemetryRepository`, `AnomalyDetectionService` y `AlertService`. De esta manera, cada clase tendría una función más concreta y sería más sencillo realizar cambios sin afectar otras partes del sistema.

Esta organización también podría facilitar las pruebas, el mantenimiento y la incorporación de nuevas funciones. Por ejemplo, se podría modificar el análisis de anomalías sin tener que cambiar directamente el almacenamiento de las mediciones.

Es importante señalar que la propuesta se basa en la arquitectura conceptual documentada de VECTRA. Para confirmar problemas específicos de implementación, sería necesario contar con el código fuente de las clases actuales y revisarlo con mayor detalle.
